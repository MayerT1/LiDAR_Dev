```python
# ---------------------------------------------
# HELPER: Filter dataframe by AOI
# ---------------------------------------------
def filter_by_aoi(df, lon_col='Longitude', lat_col='Latitude'):
    points = gpd.GeoDataFrame(df, geometry=gpd.points_from_xy(df[lon_col], df[lat_col]), crs='EPSG:4326')
    return gpd.sjoin(points, aoi_gdf, how='inner', predicate='within')
```

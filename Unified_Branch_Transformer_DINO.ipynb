```python
import torch
import torch.nn as nn
from torchvision.models.vision_transformer import vit_b_16, ViT_B_16_Weights

class EOViT(nn.Module):
    def __init__(self, input_channels=4, batch_norm=True, projection=False):
        super().__init__()

        # Load standard ViT backbone and adapt input layer
        vit = vit_b_16(weights=None)
        vit.conv_proj = nn.Conv2d(input_channels, vit.hidden_dim, kernel_size=16, stride=16)

        # Optional batch norm after input projection
        if batch_norm:
            self.batchnorm = nn.BatchNorm2d(vit.hidden_dim)
        else:
            self.batchnorm = nn.Identity()

        self.vit = vit
        self.projection = projection

        # Replace classifier head with MLP for embedding extraction
        self.mlp_head = nn.Sequential(
            nn.LayerNorm(vit.hidden_dim),
            nn.Linear(vit.hidden_dim, 1)
        )

    def forward(self, x):
        # x: [B, C, H, W]
        x = self.vit.conv_proj(x)      # Patch embedding
        x = self.batchnorm(x)          # Optional
        x = x.flatten(2).transpose(1, 2)  # Flatten to patches

        x = self.vit.encoder(x)        # Transformer encoder

        cls_token = x[:, 0]            # CLS token
        out = self.mlp_head(cls_token) # MLP regression

        return out

    def extract_cls_embeddings(self, x):
        # Returns the CLS token embeddings
        x = self.vit.conv_proj(x)
        x = self.batchnorm(x)
        x = x.flatten(2).transpose(1, 2)
        x = self.vit.encoder(x)
        cls_token = x[:, 0]
        return cls_token
```

```python
#🔷 ActivatedEOViT Block — (Wraps EOViT with Projection Head)
import torch
import torch.nn as nn

class ActivatedEOViT(nn.Module):
    def __init__(self, base_model):
        super().__init__()
        self.base_model = base_model

        # Projection head: maps CLS embedding to 128-D for DINO loss
        self.proj_head = nn.Sequential(
            nn.Linear(base_model.mlp_head[1].in_features, 256),
            nn.ReLU(),
            nn.Linear(256, 128),
        )

    def forward(self, x, return_proj=False):
        """
        Args:
            x: input tensor [B, C, H, W]
            return_proj: whether to return both embedding and projection
        Returns:
            emb: CLS token embedding
            proj (optional): projection for contrastive learning
        """
        emb = self.base_model.extract_cls_embeddings(x)

        if return_proj:
            proj = self.proj_head(emb)
            return emb, proj
        else:
            return emb
```

```python
#🔷 DinoWrapper Class — (Manages Teacher/Student Models + EMA)
import copy
import torch
import torch.nn as nn

class DinoWrapper(nn.Module):
    def __init__(self, student_model, ema_momentum=0.996):
        super().__init__()
        self.student = student_model
        self.teacher = copy.deepcopy(student_model)
        self.ema_momentum = ema_momentum

        # Disable gradients for the teacher model
        for param in self.teacher.parameters():
            param.requires_grad = False

    def forward(self, v1, v2, return_proj=True):
        """
        Args:
            v1, v2: Two views of the same image (augmented)
            return_proj: If True, return (student_proj, teacher_proj)
        Returns:
            Student and teacher projections
        """
        # Student gets both views
        _, student_proj = self.student(v1, return_proj=return_proj)

        # Teacher gets only the second view
        with torch.no_grad():
            _, teacher_proj = self.teacher(v2, return_proj=return_proj)

        return student_proj, teacher_proj

    def update_teacher(self):
        """
        EMA update: teacher = momentum * teacher + (1 - momentum) * student
        """
        for student_param, teacher_param in zip(self.student.parameters(), self.teacher.parameters()):
            teacher_param.data = self.ema_momentum * teacher_param.data + (1.0 - self.ema_momentum) * student_param.data
```

```python
#🔷 train_one_epoch_dino() — (1 Epoch of Self-Supervised DINO Training)

def train_one_epoch_dino(model, dataloader, optimizer, dino_loss_fn, device):
    """
    Train the DINO wrapper for one epoch.

    Args:
        model (DinoWrapper): Student-Teacher wrapper
        dataloader (DataLoader): DINO-style dataloader returning (view1, view2)
        optimizer (torch.optim.Optimizer): Optimizer for student model
        dino_loss_fn (DINOLoss): DINO loss function
        device (torch.device): CUDA or CPU
    Returns:
        Average loss across the epoch
    """
    model.train()
    total_loss = 0.0

    for batch_idx, (v1, v2, _, _) in enumerate(dataloader):
        v1, v2 = v1.to(device), v2.to(device)

        # Forward pass through DINO model wrapper
        student_proj, teacher_proj = model(v1, v2, return_proj=True)

        # Compute DINO loss
        loss = dino_loss_fn(student_proj, teacher_proj)

        # Backward + optimizer step (only for student)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        # Update teacher via EMA
        model.update_teacher()

        total_loss += loss.item()

        if batch_idx % 10 == 0:
            print(f"[DINO TRAIN] Batch {batch_idx} Loss: {loss.item():.4f}")

    avg_loss = total_loss / len(dataloader)
    print(f"[DINO TRAIN] Avg Loss: {avg_loss:.4f}")
    return avg_loss
```

```python
#🔷 hyperparameter_search() — (Train + Evaluate Models Over Grid)
def hyperparameter_search(eo, train_dataset, test_dataset, val_dataset, batch_size, device):
    from torch.utils.data import DataLoader
    import torch

    search_space = [
        {"batch_norm": True, "lr": 1e-4, "opt": "Adam", "epochs": 50},
        {"batch_norm": False, "lr": 1e-4, "opt": "AdamW", "epochs": 50},
        # Add more combos as needed
    ]

    best_model = None
    best_val_score = float("inf")
    best_config = None

    for config in search_space:
        print(f"\n[GRID SEARCH] batch_norm={config['batch_norm']}, lr={config['lr']}, opt={config['opt']}, epochs={config['epochs']}")

        # 1. Build EOViT base model
        base_model = EOViT(in_channels=4, batch_norm=config["batch_norm"], projection=False)
        print(f"[INFO] Initialized EOViT with batch_norm={config['batch_norm']}, projection=False for 4 input channels.")

        # 2. Wrap it with projection head
        activated_model = ActivatedEOViT(base_model)

        # 3. Wrap it with student-teacher DINO wrapper
        model = DinoWrapper(activated_model, device=device).to(device)

        # 4. Optimizer
        optimizer_cls = torch.optim.Adam if config["opt"] == "Adam" else torch.optim.AdamW
        optimizer = optimizer_cls(model.parameters(), lr=config["lr"])

        # 5. Loss
        dino_loss_fn = DINOLoss(
            out_dim=activated_model.proj_head[-1].out_features,
            ncrops=2,
            warmup_teacher_temp=0.04,
            teacher_temp=0.07,
            warmup_teacher_temp_epochs=10,
            nepochs=config["epochs"]
        ).to(device)

        # 6. Dataloaders
        train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
        val_loader = DataLoader(val_dataset, batch_size=batch_size)

        # 7. Train
        for epoch in range(config["epochs"]):
            print(f"[EPOCH {epoch+1}/{config['epochs']}]")
            train_one_epoch_dino(model, train_loader, optimizer, dino_loss_fn, device)

        # 8. Evaluate on validation set
        val_loss = evaluate(model, val_loader, dino_loss_fn, device)
        print(f"[VAL LOSS] {val_loss:.4f}")

        # 9. Keep best model
        if val_loss < best_val_score:
            best_val_score = val_loss
            best_model = model
            best_config = config

    print(f"\n[SEARCH DONE] Best config: {best_config}, val_loss: {best_val_score:.4f}")
    return best_model
```

```python
import torch
import torch.nn.functional as F

class DINOLoss(torch.nn.Module):
    def __init__(self, out_dim, ncrops, warmup_teacher_temp=0.04, teacher_temp=0.07, warmup_teacher_temp_epochs=30, nepochs=100):
        super().__init__()
        self.out_dim = out_dim
        self.ncrops = ncrops
        self.teacher_temp = teacher_temp
        self.warmup_teacher_temp = warmup_teacher_temp
        self.warmup_teacher_temp_epochs = warmup_teacher_temp_epochs
        self.nepochs = nepochs
        self.register_buffer("center", torch.zeros(1, out_dim))
        self.current_epoch = 0

    def forward(self, student_output, teacher_output):
        """
        student_output: (batch_size * ncrops, out_dim)
        teacher_output: (batch_size * ncrops, out_dim)
        """
        student_out = student_output.chunk(self.ncrops)
        teacher_out = teacher_output.chunk(self.ncrops)

        loss = 0
        temp = self.get_teacher_temp()
        for iq, q in enumerate(teacher_out):
            for v in range(len(student_out)):
                if v == iq:
                    continue
                # Teacher output centered and temp scaled
                t = F.softmax((q - self.center) / temp, dim=-1).detach()
                s = F.log_softmax(student_out[v] / 0.1, dim=-1)
                loss += -(t * s).sum(dim=-1).mean()

        loss /= (self.ncrops - 1) * self.ncrops
        self.update_center(teacher_output)
        return loss

    def get_teacher_temp(self):
        # Linear warmup
        if self.current_epoch < self.warmup_teacher_temp_epochs:
            return self.warmup_teacher_temp + (self.teacher_temp - self.warmup_teacher_temp) * self.current_epoch / self.warmup_teacher_temp_epochs
        else:
            return self.teacher_temp

    def update_center(self, teacher_output):
        batch_center = torch.mean(teacher_output, dim=0, keepdim=True)
        self.center = 0.9 * self.center + 0.1 * batch_center

    def set_epoch(self, epoch):
        self.current_epoch = epoch
```

```python
class DinoModelWrapper(nn.Module):
    def __init__(self, base_model, momentum=0.996):
        super().__init__()
        self.student = base_model
        self.teacher = copy.deepcopy(base_model)
        self.momentum = momentum
        # Teacher is not updated by gradient
        for param in self.teacher.parameters():
            param.requires_grad = False

    def forward(self, x, return_proj=False):
        return self.student(x, return_proj=return_proj)

    def update_teacher(self):
        # EMA update of teacher params
        with torch.no_grad():
            for student_param, teacher_param in zip(self.student.parameters(), self.teacher.parameters()):
                teacher_param.data = self.momentum * teacher_param.data + (1 - self.momentum) * student_param.data
```

```python
# --- Dataset ---
class EODataset(Dataset):
    def __init__(self, eo_name, data_dir, split, target_df, patch_size=32, transform=None):
        self.eo_name = eo_name
        self.data_dir = data_dir
        self.split = split
        self.patch_size = patch_size
        self.target_df = target_df
        self.transform = transform

        # Match only .tif files that are in the dataframe
        all_files = sorted([
            f for f in os.listdir(data_dir)
            if f.startswith(f"{eo_name}_{split}_") and f.endswith(".tif")
        ])
        target_files = set(target_df['filename'])
        self.files = [f for f in all_files if f in target_files]

        print(f"[INFO] {split.upper()} dataset for {eo_name} initialized with {len(self.files)} tiles.")

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        filename = self.files[idx]
        filepath = os.path.join(self.data_dir, filename)

        try:
            with rasterio.open(filepath) as src:
                array_resized = src.read(
                    out_shape=(src.count, self.patch_size, self.patch_size),
                    resampling=rasterio.enums.Resampling.bilinear
                )
                bounds = src.bounds
                x_center = (bounds.left + bounds.right) / 2
                y_center = (bounds.top + bounds.bottom) / 2
        except Exception as e:
            raise RuntimeError(f"Error reading {filepath}: {e}")

        img_tensor = torch.from_numpy(array_resized).float()  # [C, H, W]

        # Apply DINO-style dual augmentations for training
        if self.transform is not None and self.split == 'train':
            v1 = self.transform(img_tensor.unsqueeze(0)).squeeze(0)
            v2 = self.transform(img_tensor.unsqueeze(0)).squeeze(0)
        else:
            v1 = v2 = img_tensor

        # Lookup target
        row = self.target_df[self.target_df['filename'] == filename]
        if len(row) == 0:
            raise ValueError(f"No target value found for file: {filename}")
        target = row['avg_pai'].values[0]
        target_tensor = torch.tensor(target, dtype=torch.float32)

        # Spatial coordinates
        spatial_tensor = torch.tensor([x_center, y_center], dtype=torch.float32)

        return v1, v2, target_tensor, spatial_tensor
```

```python
import torch
import torch.nn as nn

class EOViT(nn.Module):
    def __init__(
        self, in_channels, patch_size=16, emb_dim=64, num_heads=4,
        num_layers=2, patch_dim=32, batch_norm=True, use_projection=False, projection_dim=256
    ):
        super().__init__()
        self.batch_norm = batch_norm
        self.use_projection = use_projection

        # Input scaling and normalization
        self.input_scaler = nn.Conv2d(in_channels, in_channels, kernel_size=1, stride=1, padding=0, bias=True)
        self.input_norm = nn.BatchNorm2d(in_channels) if batch_norm else nn.Identity()

        # Patch embedding
        self.patch_embed = nn.Conv2d(in_channels, emb_dim, kernel_size=patch_size, stride=patch_size)

        num_patches = (patch_dim // patch_size) ** 2
        self.cls_token = nn.Parameter(torch.randn(1, 1, emb_dim))
        self.pos_embed = nn.Parameter(torch.randn(1, num_patches + 1, emb_dim))

        # Transformer encoder
        encoder_layer = nn.TransformerEncoderLayer(d_model=emb_dim, nhead=num_heads, batch_first=True)
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)

        # Regression head for supervised tasks
        self.mlp_head = nn.Sequential(
            nn.LayerNorm(emb_dim),
            nn.Linear(emb_dim, 1)
        )

        # Optional projection head for DINO
        if use_projection:
            self.projection_head = nn.Sequential(
                nn.Linear(emb_dim, emb_dim),
                nn.GELU(),
                nn.Linear(emb_dim, projection_dim)
            )

        print(f"[INFO] Initialized EOViT with batch_norm={batch_norm}, projection={use_projection} for {in_channels} input channels.")

    def forward(self, x, return_cls=False, return_proj=False):
        x = self.input_scaler(x)
        x = self.input_norm(x)
        x = self.patch_embed(x).flatten(2).transpose(1, 2)

        cls_tokens = self.cls_token.expand(x.size(0), -1, -1)
        x = torch.cat((cls_tokens, x), dim=1)
        x = x + self.pos_embed
        x = self.transformer(x)

        cls_embedding = x[:, 0]

        if return_proj and self.use_projection:
            return self.projection_head(cls_embedding)
        elif return_cls:
            return cls_embedding
        else:
            return self.mlp_head(cls_embedding).squeeze(1)

    def extract_cls_embeddings(self, x):
        return self.forward(x, return_cls=True)
```

```python
def train_one_epoch_dino(model, dataloader, optimizer, dino_loss_fn, device):
    model.train()
    total_loss = 0.0

    for batch_idx, (v1, v2, _, _) in enumerate(dataloader):
        v1, v2 = v1.to(device), v2.to(device)

        # Call model separately on each view to get embeddings and projections
        student_emb, student_proj = model(v1, return_proj=True)
        teacher_emb, teacher_proj = model(v2, return_proj=True)

        # Compute loss on projections
        loss = dino_loss_fn(student_proj, teacher_proj)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        model.update_teacher()

        total_loss += loss.item()
        if batch_idx % 10 == 0:
            print(f"[DINO TRAIN] Batch {batch_idx} Loss: {loss.item():.4f}")

    return total_loss / len(dataloader)
```

```python
def evaluate(model, dataloader, criterion, device):
    model.eval()
    running_loss = 0.0
    preds, trues = [], []

    with torch.no_grad():
        for x, y, _ in dataloader:
            x, y = x.to(device), y.to(device)

            # Forward pass through student model only
            out = model.student(x)  # assuming .student is an EOViT instance with regression head
            loss = criterion(out, y)

            running_loss += loss.item()
            preds.append(out.cpu().numpy())
            trues.append(y.cpu().numpy())

    avg_loss = running_loss / len(dataloader)
    preds = np.concatenate(preds)
    trues = np.concatenate(trues)
    rmse = np.sqrt(mean_squared_error(trues, preds))

    print(f"[EVAL] Avg Loss: {avg_loss:.4f} RMSE: {rmse:.4f}")
    return avg_loss, rmse
```

```python
# --- Hyperparameter grid search ---
def hyperparameter_search(eo, train_dataset, test_dataset, batch_size=8, epochs=10, device='cpu'):
    print(f"[INFO] Starting hyperparameter search for {eo}")
    best_rmse = float('inf')
    best_params = None
    best_model = None

    batch_norm_options = [True]  # e.g., [True, False]
    lr_options = [1e-4]          # e.g., [1e-3, 5e-4, 1e-4]
    epoch_options = [50]         # e.g., [20, 30, 50]
    optimizers = ['Adam']        # e.g., ['Adam', 'AdamW']

    train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
    test_loader = DataLoader(test_dataset, batch_size=batch_size)

    for batch_norm in batch_norm_options:
        for lr in lr_options:
            for opt_name in optimizers:
                for n_epochs in epoch_options:
                    print(f"\n[GRID SEARCH] batch_norm={batch_norm}, lr={lr}, opt={opt_name}, epochs={n_epochs}")

                    base_model = EOViT(
                        in_channels=bands_per_cov[eo],
                        patch_size=16,
                        emb_dim=64,
                        num_heads=4,
                        num_layers=2,
                        patch_dim=train_dataset.patch_size,
                        batch_norm=batch_norm
                    ).to(device)

                    model = ActivatedEOViT(base_model).to(device)

                    optimizer = torch.optim.Adam(model.parameters(), lr=lr) if opt_name == 'Adam' \
                        else torch.optim.AdamW(model.parameters(), lr=lr)

                    dino_loss_fn = DINOLoss(  # Make sure DINOLoss is properly defined elsewhere
                        out_dim=base_model.mlp_head[1].out_features,
                        ncrops=2
                    ).to(device)

                    train_losses, test_losses = [], []
                    for epoch in range(n_epochs):
                        print(f"[EPOCH {epoch+1}/{n_epochs}]")
                        train_loss = train_one_epoch_dino(model, train_loader, optimizer, dino_loss_fn, device)
                        test_loss, test_rmse = evaluate(model, test_loader, nn.MSELoss(), device)

                        train_losses.append(train_loss)
                        test_losses.append(test_loss)

                    # Plotting
                    plt.figure()
                    plt.plot(range(1, n_epochs + 1), train_losses, label='Train Loss')
                    plt.plot(range(1, n_epochs + 1), test_losses, label='Test Loss')
                    plt.title(f'{eo} Loss\nBN={batch_norm}, LR={lr}, OPT={opt_name}')
                    plt.xlabel('Epoch')
                    plt.ylabel('Loss')
                    plt.legend()
                    plt.show()

                    # Save best model
                    if test_rmse < best_rmse:
                        best_rmse = test_rmse
                        best_params = {
                            'batch_norm': batch_norm,
                            'lr': lr,
                            'epochs': n_epochs,
                            'optimizer': opt_name
                        }
                        best_model = model

    print(f"[INFO] Best params for {eo}: {best_params} with RMSE={best_rmse:.4f}")
    return best_model, best_params
```

```python
# --- Extract embeddings and spatial info for val set ---
def extract_embeddings_and_spatial(model, dataloader, device):
    model.eval()
    all_embeddings, all_targets, all_coords = [], [], []

    with torch.no_grad():
        for x, y, coord in dataloader:
            x = x.to(device)
            emb = model.extract_cls_embeddings(x).cpu()
            all_embeddings.append(emb)
            all_targets.append(y.cpu())   # move to CPU to keep consistent
            all_coords.append(coord.cpu()) # same here, in case coords are tensors on GPU

    embeddings = torch.cat(all_embeddings, dim=0).numpy()
    targets = torch.cat(all_targets, dim=0).numpy()
    coords = torch.cat(all_coords, dim=0).numpy()

    print(f"[INFO] Extracted {len(embeddings)} embeddings with spatial info for validation.")
    return embeddings, targets, coords
```

```python
def stack_embeddings_for_fusion(embedding_dict):
    eo_keys = sorted(embedding_dict.keys())
    emb_list = [embedding_dict[eo]['embeddings'] for eo in eo_keys]

    # Make sure all have same number of samples and coords
    for i in range(1, len(emb_list)):
        assert emb_list[i].shape[0] == emb_list[0].shape[0], f"Mismatch in #samples between {eo_keys[0]} and {eo_keys[i]}"

    # [N, D_total] by stacking along feature dimension
    stacked_embeddings = np.concatenate(emb_list, axis=1)
    coords = embedding_dict[eo_keys[0]]['coords']
    targets = embedding_dict[eo_keys[0]]['targets']

    return stacked_embeddings, coords, targets
```

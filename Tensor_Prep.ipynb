```python
import os
import folium
import rasterio
from rasterio.warp import transform_bounds
from glob import glob
from shapely.geometry import box, Point, mapping
import geopandas as gpd
import pandas as pd

# ----------------- CONFIG -----------------
RASTER_DIR = '/content/drive/MyDrive/PhD_Main_Folder/Geo_Data/Sewanee_DL_Project/data'
GEDI_CSV = '/content/drive/MyDrive/PhD_Main_Folder/Geo_Data/Sewanee_DL_Project/target_data/filtered_csvs/GEDI_filtered_merged.csv'

COVARIATE_COLORS = {
    'dem': 'gray',
    'ls': 'green',
    'ndvi': 'purple',
    'savi': 'blue',
    'evi': 'orange',
    'slope': 'red',
    'aspect': 'brown',
    'elevation': 'black',
    'default': 'pink'
}

# ----------------- Load GEDI Targets -----------------
gedi_df = pd.read_csv(GEDI_CSV)

lat_col_candidates = ['latitude', 'lat', 'y']
lon_col_candidates = ['longitude', 'lon', 'x']

lat_col = next((col for col in lat_col_candidates if col in gedi_df.columns), None)
lon_col = next((col for col in lon_col_candidates if col in gedi_df.columns), None)

if not lat_col or not lon_col:
    raise ValueError(f"Latitude/Longitude columns not found. Expected one of {lat_col_candidates} and {lon_col_candidates}")

gedi_gdf = gpd.GeoDataFrame(
    gedi_df,
    geometry=gpd.points_from_xy(gedi_df[lon_col], gedi_df[lat_col]),
    crs='EPSG:4326'
)

# Create square buffer (~25m = ~0.000225 degrees at equator)
gedi_gdf['geometry'] = gedi_gdf.geometry.buffer(0.000225, cap_style=3)
gedi_gdf['block_id'] = gedi_gdf['geometry'].centroid.apply(lambda pt: f"{int(pt.x*1e4)}_{int(pt.y*1e4)}")

# ----------------- Load Raster Bounds -----------------
tif_files = glob(os.path.join(RASTER_DIR, '**/*.tif'), recursive=True)

raster_records = []
for tif in tif_files:
    try:
        with rasterio.open(tif) as src:
            bounds = transform_bounds(src.crs, 'EPSG:4326', *src.bounds)
            cov = next((c for c in COVARIATE_COLORS if c in tif.lower()), 'default')
            color = COVARIATE_COLORS[cov]
            geom = box(*bounds)
            block_id = os.path.basename(tif).split('_')[-1].replace('.tif', '')
            raster_records.append({
                'geometry': geom,
                'block_id': block_id,
                'covariate': cov,
                'resolution': src.res[0],
                'color': color,
                'filename': os.path.basename(tif)
            })
    except Exception as e:
        print(f"[WARN] Skipping {tif}: {e}")

raster_gdf = gpd.GeoDataFrame(raster_records, crs='EPSG:4326')

#----------------- Initialize Map -----------------
center = gedi_gdf.geometry.unary_union.centroid
m = folium.Map(location=[center.y, center.x], zoom_start=13, tiles="CartoDB positron")

# ----------------- Add GEDI Layer -----------------
for _, row in gedi_gdf.iterrows():
    coords = list(mapping(row.geometry)['coordinates'][0])
    folium.Polygon(
        locations=[(lat, lon) for lon, lat in coords],
        color='cyan',
        fill=True,
        fill_opacity=0.4,
        tooltip=f"<b>GEDI</b><br>Block: {row['block_id']}"
    ).add_to(m)

# ----------------- Add Raster Layers (Grouped by Covariate) -----------------
for covariate in raster_gdf['covariate'].unique():
    sub_gdf = raster_gdf[raster_gdf['covariate'] == covariate]
    for _, row in sub_gdf.iterrows():
        coords = list(mapping(row.geometry)['coordinates'][0])
        folium.Polygon(
            locations=[(lat, lon) for lon, lat in coords],
            color=row['color'],
            weight=1,
            fill=True,
            fill_opacity=0.2,
            tooltip=f"""
                <b>{covariate.upper()}</b><br>
                Block ID: {row['block_id']}<br>
                Resolution: {row['resolution']}<br>
                File: {row['filename']}
            """
        ).add_to(m)

# ----------------- Show Map -----------------
m
```

Average GEDI values per 32x32 block

```python
# ----------------- Ensure Same CRS -----------------
gedi_gdf = gedi_gdf.to_crs('EPSG:4326')
raster_gdf = raster_gdf.to_crs('EPSG:4326')

# ----------------- Spatial Join -----------------
# Each GEDI point gets linked to overlapping raster tile
joined = gpd.sjoin(gedi_gdf, raster_gdf, how='inner', predicate='intersects')

# ----------------- Compute Per-Tile Averages -----------------
target_cols = ['pai', 'fhd_normal', 'rh100']

# Check original target columns exist
for col in target_cols:
    if col not in joined.columns:
        raise ValueError(f"Missing required column '{col}' in joined GeoDataFrame")

# Group by raster polygon index and compute averages
avg_df = (
    joined
    .groupby('index_right')[target_cols]
    .mean()
    .rename(columns={
        'pai': 'avg_pai',
        'fhd_normal': 'avg_fhd_normal',
        'rh100': 'avg_rh100'
    })
    .reset_index()
)

# ----------------- Merge Averages with Raster Polygons -----------------
# Reset index to get 'index_right' for joining
raster_meta = raster_gdf.reset_index().rename(columns={'index': 'index_right'})

# Merge averages onto raster polygons
raster_with_avg = raster_meta.merge(avg_df, on='index_right', how='left')

# ----------------- Finalize Output -----------------
# raster_with_avg is a GeoDataFrame with one polygon per raster tile,
# and averaged GEDI target values per polygon.

# Select relevant columns (add more if needed)
final_cols = [
    'geometry', 'block_id', 'covariate', 'filename', 'resolution',
    'avg_pai', 'avg_fhd_normal', 'avg_rh100'
]

final_df = raster_with_avg[final_cols].copy()

# ----------------- Output -----------------
print("✅ Final raster polygons with averaged GEDI target values:")
print(final_df.head())
print(f"\n✅ Total raster polygons with GEDI data: {len(final_df)}")
```

```python
import os
import numpy as np
import rasterio
import torch
from tqdm import tqdm

# --- Clean up paths ---
final_df['filename'] = final_df['filename'].str.strip()
final_df['full_path'] = final_df['filename'].apply(lambda x: os.path.join(RASTER_FOLDER, x))

# --- Drop rows with missing target values ---
final_df = final_df.dropna(subset=['avg_pai', 'avg_fhd_normal', 'avg_rh100']).reset_index(drop=True)

# --- Prepare lists to collect data ---
X_list = []
y_list = []

# --- Iterate through all rows ---
for idx, row in tqdm(final_df.iterrows(), total=len(final_df)):
    try:
        # Load the raster
        with rasterio.open(row['full_path']) as src:
            data = src.read()  # shape: (bands, height, width)

        # Optional: check for NaNs or zero shape
        if np.any(np.isnan(data)) or data.size == 0:
            continue

        # Store covariate (as torch tensor)
        X_list.append(torch.tensor(data, dtype=torch.float32))

        # Store targets
        target = torch.tensor([row['avg_pai'], row['avg_fhd_normal'], row['avg_rh100']], dtype=torch.float32)
        y_list.append(target)

    except Exception as e:
        print(f"Error reading {row['full_path']}: {e}")
        continue

# --- Stack tensors into final dataset ---
X = torch.stack(X_list)  # Shape: (N, C, H, W)
y = torch.stack(y_list)  # Shape: (N, 3)

print(f"✅ Done! Covariates shape: {X.shape}, Targets shape: {y.shape}")
```

```python
for col in raster_with_targets.columns:
    print(col)
import os
import torch
import rasterio
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from rasterio.windows import Window
from glob import glob

# ------------------------ Config ------------------------
TARGET_CSV = '/content/drive/MyDrive/PhD_Main_Folder/Geo_Data/Sewanee_DL_Project/target_data/filtered_csvs/GEDI_filtered_merged.csv'
RASTER_FOLDER = '/content/drive/MyDrive/PhD_Main_Folder/Geo_Data/Sewanee_DL_Project/data'
PATCH_SIZE = 32
RESOLUTIONS = [10, 20, 30]
SPLIT = 'train'

# ------------------------ Step 1: Discover Available Raster Blocks ------------------------
def extract_block_ids_from_rasters(res):
    files = glob(os.path.join(RASTER_FOLDER, f"*_{res}m_*.tif"))
    block_ids = set()
    for f in files:
        parts = os.path.basename(f).split('_')
        if len(parts) >= 3:
            try:
                block_id = parts[-1].replace('.tif', '')
                block_ids.add(block_id)
            except:
                continue
    return block_ids

# Get common block_ids across all resolutions
available_blocks_per_res = [extract_block_ids_from_rasters(res) for res in RESOLUTIONS]
common_raster_blocks = set.intersection(*available_blocks_per_res)
print(f"✅ {len(common_raster_blocks)} blocks with all resolutions available.")

# ------------------------ Step 2: Filter Target CSV ------------------------
df = pd.read_csv(TARGET_CSV)
df_split = df[df['split'] == SPLIT]
df_split = df_split[df_split['block_id'].astype(str).isin(common_raster_blocks)]
grouped = df_split.groupby('block_id')

print(f"✅ {len(df_split)} samples remain after filtering non-overlapping blocks.")

# ------------------------ Helper Functions ------------------------
def get_raster_paths(block_id):
    paths = {}
    for res in RESOLUTIONS:
        matches = sorted(glob(os.path.join(RASTER_FOLDER, f"*_{res}m_*{block_id}.tif")))
        if matches:
            paths[res] = matches[0]
        else:
            return None
    return paths

def extract_patch(dataset, lat, lon, size):
    try:
        row, col = dataset.index(lon, lat)
        half_size = size // 2
        window = Window(col - half_size, row - half_size, size, size)
        patch = dataset.read(window=window)
        if patch.shape[1] != size or patch.shape[2] != size:
            raise ValueError("Patch outside bounds")
        return patch
    except Exception as e:
        raise RuntimeError(f"Failed to extract patch at {lat}, {lon}: {e}")

# ------------------------ Step 3: Process Data ------------------------
X_data = []
y_data = []

for block_id, group in grouped:
    raster_paths = get_raster_paths(block_id)
    if raster_paths is None:
        continue

    try:
        raster_datasets = {res: rasterio.open(path) for res, path in raster_paths.items()}
    except Exception as e:
        print(f"[ERROR] Could not open rasters for block {block_id}: {e}")
        continue

    for _, row in group.iterrows():
        lat, lon = row['lat'], row['lon']
        try:
            patches = []
            for res in RESOLUTIONS:
                patch = extract_patch(raster_datasets[res], lat, lon, PATCH_SIZE)
                patch_tensor = torch.tensor(patch, dtype=torch.float32)
                patches.append(patch_tensor)
            multi_res_patch = torch.cat(patches, dim=0)
            X_data.append(multi_res_patch)
            y_data.append(torch.tensor(row['pai'], dtype=torch.float32))
        except Exception as e:
            print(f"[SKIP] {e}")
            continue

    for ds in raster_datasets.values():
        ds.close()

# ------------------------ Step 4: Create Tensors ------------------------
if not X_data:
    raise RuntimeError("No valid patches loaded. Ensure there is spatial overlap between rasters and targets.")

X_tensor = torch.stack(X_data)
y_tensor = torch.stack(y_data)

print("✅ Done loading data.")
print("Input tensor shape:", X_tensor.shape)
print("Target tensor shape:", y_tensor.shape)

# ------------------------ Visualize ------------------------
idx = 0
sample = X_tensor[idx]
n_channels = sample.shape[0]
fig, axes = plt.subplots(1, n_channels, figsize=(n_channels * 3, 3))
for i in range(n_channels):
    axes[i].imshow(sample[i].numpy(), cmap='viridis')
    axes[i].set_title(f'Ch {i+1}')
    axes[i].axis('off')
plt.suptitle(f"Sample PAI: {y_tensor[idx].item():.2f}")
plt.tight_layout()
plt.show()
```
